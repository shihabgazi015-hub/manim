# Animating a circle

````python
%load_ext manim
````

````python
%%manim -pql CreateCircle

from manim import *


class CreateCircle(Scene):
    def construct(self):
        circle = Circle()  # create a circle
        circle.set_fill(PINK, opacity=0.8)  # set the color and transparency
        self.play(Create(circle))  # show the circle on screen
````

## Explanation

#### Importing the Library
```python
from manim import *
```

#### Constructing the Scene
Most of the time, the script for an animation is entirely contained within the `construct()` method of a `Scene` class. Inside this method, objects can be created, displayed on screen, and animated.

```python
class CreateCircle(Scene):
    def construct(self):
        [...]
```

#### Creating the Circle
```python
circle = Circle() # Create a circle object
circle.set_fill(PINK, opacity=0.8) # Set the color and transparency
```

#### Animating the Circle
```python
self.play(Create(circle)) # Animate the circle's creation on screen
```

> [!TIP]\
> Custom animation class can be named anything you like (e.g., `CreateCircle`), but it must inherit from Manim's base `Scene` class.


# Transforming a square into a circle

````python
%%manim -pql SquareToCircle

class SquareToCircle(Scene):
    def construct(self):
        circle = Circle()
        circle.set_fill(PINK, opacity = 0.75)

        square = Square()
        square.rotate(PI/4) # rotate a cirtain amount in this case pi/4

        self.play(Create(square)) # animating the creation of the square
        self.wait(1)
        self.play(Transform(square, circle)) # interpolate the square into the circle
        self.wait(1)
        self.play(FadeOut(square)) # fade out animating
````

The `Transform(a, b)` method transforms the object *a* to object *b* but, mainly the object 'a' is stored in the computer memory. On the other hand `ReplacementTransform(a, b)` method **replaces** the object *a* by object *b*

````python
%%manim -pql RplacingSquareByCircle

class RplacingSquareByCircle(Scene):
    def construct(self):
        circle = Circle()
        circle.set_fill(PINK, opacity = 0.75)

        square = Square()
        square.rotate(PI/4) # rotate a cirtain amount in this case pi/4

        self.play(Create(square)) # animating the creation of the square
        self.wait(1)
        self.play(ReplacementTransform(square, circle)) # replacing the square by the circle
        self.wait(1)
        self.play(FadeOut(circle)) # fade out the circle
````

# Positioning Mobject S

````python
%%manim -pql SquareAndCircle

class SquareAndCircle(Scene):
    def construct(self):
        circle = Circle()  # create a circle
        circle.set_fill(PINK, opacity=0.5)  # set the color and transparency

        square = Square()  # create a square
        square.set_fill(BLUE, opacity=0.5)  # set the color and transparency

        square.next_to(circle, RIGHT, buff=0.5)  # set the position
        self.play(Create(circle), Create(square))  # show the shapes on screen
        self.wait(1.5)
````

<video src="media/videos/tutorials/480p15/SquareAndCircle.mp4" controls width="100%"></video>

`next_to` is a `Mobject` method for positioning Mobjects.\
`circle` is the method’s **first** argument. The **second** argument is used to specify the **direction** the Mobject is placed relative to the reference point.\
`buff=0.5` applied a small distance buffer between the two objects.
