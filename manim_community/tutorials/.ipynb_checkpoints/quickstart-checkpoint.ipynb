# Animating a circle

````python
%load_ext manim
````

````python
%%manim -pql CreateCircle

from manim import *


class CreateCircle(Scene):
    def construct(self):
        circle = Circle()  # create a circle
        circle.set_fill(PINK, opacity=0.8)  # set the color and transparency
        self.play(Create(circle))  # show the circle on screen
````

## Explanation

#### Importing the Library
```python
from manim import *
```

#### Constructing the Scene
Most of the time, the script for an animation is entirely contained within the `construct()` method of a `Scene` class. Inside this method, objects can be created, displayed on screen, and animated.

```python
class CreateCircle(Scene):
    def construct(self):
        [...]
```

#### Creating the Circle
```python
circle = Circle() # Create a circle object
circle.set_fill(PINK, opacity=0.8) # Set the color and transparency
```

#### Animating the Circle
```python
self.play(Create(circle)) # Animate the circle's creation on screen
```

> [!TIP]\
> Custom animation class can be named anything you like (e.g., `CreateCircle`), but it must inherit from Manim's base `Scene` class.


# Transforming a square into a circle

````python
class SquareToCircle(Scene):
    def construct(self):
        circle = Circle()  # create a circle
        circle.set_fill(PINK, opacity=0.5)  # set color and transparency

        square = Square()  # create a square
        square.rotate(PI / 4)  # rotate a certain amount

        self.play(Create(square))  # animate the creation of the square
        self.play(Transform(square, circle))  # interpolate the square into the circle
        self.play(FadeOut(square))  # fade out animation
````
